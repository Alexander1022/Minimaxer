```
Критерий:
  5x + 3y

Ограничения:
  2x + y <= 10
  x + y <= 7
  x, y >= 0
```

````python
problem = pulp.LpProblem("Example", pulp.LpMaximize)

x = pulp.LpVariable("x", lowBound=0)
y = pulp.LpVariable("y", lowBound=0)

problem += 5 * x + 3 * y

problem += 2 * x + y <= 10
problem += x + y <= 7

problem.solve()
````
