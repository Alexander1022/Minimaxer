```python
# алтернативи x критерии
matrix = np.array([
    [200, 8],
    [300, 9],
    [150, 6]
])

# тегла за всеки критерий, сборът им трябва да е 1.0
w = np.array([0.5, 0.5])

# True ако критерият се максимизира, False ако се минимизира
benefit = np.array([False, True])

# Нормализация на матрицата (Vector Normalization)
norm_matrix = matrix / np.sqrt((matrix**2).sum(axis=0))

# Претегляне с теглата
weighted_matrix = norm_matrix * w

# Определяне на Идеалното (PIS) и Анти-идеалното (NIS) решение
pis = np.where(benefit, weighted_matrix.max(axis=0), weighted_matrix.min(axis=0))
nis = np.where(benefit, weighted_matrix.min(axis=0), weighted_matrix.max(axis=0))

# Изчисление на Евклидовото разстояние до PIS и NIS
s_pos = np.sqrt(((weighted_matrix - pis)**2).sum(axis=1))
s_neg = np.sqrt(((weighted_matrix - nis)**2).sum(axis=1))

# Изчисляване на относителната близост
closeness = s_neg / (s_pos + s_neg)
```

```python
# близост
closeness
```

```python
# сорт по близост
np.argsort(closeness)[::-1]
```
